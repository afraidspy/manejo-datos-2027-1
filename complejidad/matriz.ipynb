```python
def imprimir_matriz_con_conteo(matriz):
    n = len(matriz)
    operaciones = 0
    for i in range(n):
        for j in range(n):
            operaciones += 1
            print(f"Operación {operaciones}: matriz[{i}][{j}] = {matriz[i][j]}")
        print()
    print(f"Total de operaciones: {operaciones}")


matriz = [
    [30, 10, 20],
    [15, 12, 8],
    [7, 22, 5]
]

imprimir_matriz_con_conteo(matriz)
```
