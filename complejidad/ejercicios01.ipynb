# Complejidad Computacional

El objetivo de esta actividad es desarrollar la capacidad de **analizar algoritmos y determinar su complejidad computacional**.

A través de una serie de ejercicios, se deberá observar cuidadosamente cada fragmento de código para:

1. **Interpretar qué realiza el algoritmo**, sin depender del nombre de la función.
2. **Analizar el número de operaciones** que se ejecutan en función del tamaño de la entrada.
3. **Identificar la complejidad temporal**, utilizando la notación Big-O.
4. **Reconocer patrones comunes**, como complejidades `O(1)`, `O(log n)`, `O(n)`, `O(n log n)`, `O(n²)` y `O(2ⁿ)`.
5. **Analizar estructuras iterativas y recursivas**, comprendiendo cómo afectan al rendimiento de un algoritmo.

La finalidad no es únicamente elegir la respuesta correcta, sino **justificar el razonamiento utilizado para obtenerla**.

### Ejercicio 03

```python
def misterio(lista):
    for i in range(len(lista)):
        for j in range(len(lista)):
            print(lista[i] + lista[j])
```

### Ejercicio 04

```python
def misterio(lista):
    i = 1
    n = len(lista)

    while i < n:
        print(lista[i])
        i *= 2
```

### Ejercicio 05

```python
def misterio(lista):
    n = len(lista)

    for i in range(n):
        j = 1

        while j < n:
            print(lista[i] + lista[j])
            j *= 2
```

### Ejercicio 06

```python
def misterio(lista):
    n = len(lista)

    for i in range(n):
        j = i

        while j < n:
            print(lista[j])
            j += 1
```

### Ejercicio 07

```python
def misterio(n):
    if n <= 1:
        return 1

    return misterio(n - 1) + misterio(n - 1)
```

### Ejercicio 08

```python
def misterio(n):
    if n <= 1:
        return 1

    return misterio(n / 2) + misterio(n / 2)
```

### Ejercicio 09

```python
def misterio(n):
    if n <= 1:
        return 1

    x = misterio(n / 2)
    return x + x
```

### Ejercicio 10

```python
def misterio(n):
    if n <= 1:
        return 1

    for i = 1 to n:
        print(i)

    return misterio(n / 2)
```
